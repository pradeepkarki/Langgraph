```python
from typing import TypedDict
from langgraph.graph import StateGraph,add_messages
from langchain_community.chat_models import ChatOllama
from IPython.display import Image, display, JSON
```

```python

class AgentState(TypedDict):
    input: str
    messages: list[str]
```

```python
def greeting_node(state: AgentState) -> AgentState:
    state["output"] = state["input"] + ",you are doing an amazing job learning LangGraph!"
    state["messages"].append(state["output"])
    return state


def greeting_node2(state: AgentState) -> AgentState:
    state["output"] = state["input"] + ",node 2"
    state["messages"].append(state["output"])
    return state
```

```python
graph = StateGraph(AgentState)


graph.add_node("node1",greeting_node)
graph.add_node("node2",greeting_node2)
graph.add_edge("node1", "node2")


graph.set_entry_point("node1")
graph.set_finish_point("node2")


greeting_app = graph.compile()
```

```python
greeting_app.invoke({"input": "Hi", "messages": [], "output": ""})
```
