```python
from typing import TypedDict

from langgraph.graph import StateGraph,add_messages
from langchain_community.chat_models import ChatOllama
```

```python
class AgentState(TypedDict):
    name : str
    age : str
    skills : list[str]
    final : str
```

```python
def first_node(state: AgentState) -> AgentState:
    state["final"] = f"Hi, {state['name']}, welcome to the Team !"
    return state


def second_node(state: AgentState) -> AgentState:
    state["final"] = state["final"] + f" You are {state['age']} years old !"
    return state

def skill_node(state: AgentState) -> AgentState:
    state["final"] = state["final"] + f" You have the following skills: {', '.join(state['skills'])} !"
    return state
```

```python
graph = StateGraph(AgentState)

#create nodes
graph.add_node("node1", first_node)
graph.add_node("node2", second_node)
graph.add_node("node3", skill_node)
#edges
graph.add_edge("node1", "node2")
graph.add_edge("node2", "node3")

#starting and end points
graph.set_entry_point("node1")
graph.set_finish_point("node3")


#compile graph

app = graph.compile()

#invoke graph

answers = app.invoke({"name": "Pradeep", "age": "30", "final": "","skills": ["Python", "LangChain", "LangGraph"]})





```
