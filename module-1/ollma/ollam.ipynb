```python
from typing import TypedDict
from pprint import pprint

from langgraph.graph import StateGraph
from langchain_community.chat_models import ChatOllama
from IPython.display import Image, display, JSON

# Use local Ollama model
llm = ChatOllama(model="qwen3:0.6b")     

# Define state
class State(TypedDict):
    input: str
    output: str

```

```python
# Node function
def chatbot(state):
    response = llm.invoke(state["input"])
    pprint(response)

    display(JSON(response.model_dump(), expanded=True))
    #display response in json
    response_json = response
    print("LLM response (JSON):", response_json)
    
    state["output"] = response
    print("User input:", state["input"])
    print("LLM response:", state["output"])
    print("Chatbot response:", response)
    return {"output": response.content}

# Build graphs
graph = StateGraph(State)

graph.add_node("chatbot", chatbot)
graph.set_entry_point("chatbot")
graph.set_finish_point("chatbot")

app = graph.compile()

# Run
result = app.invoke({"input": "Explain python in simple terms."})
print(result["output"])

```

```python

# Show
display(Image(app.get_graph(xray=True).draw_mermaid_png()))
```
